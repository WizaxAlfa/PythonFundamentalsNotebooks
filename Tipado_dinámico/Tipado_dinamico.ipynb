# OBJETOS INMUTABLES
#### Contenido no modificable despues de su creación
Content that cannot be modified after it is created

```python
a = 3
b = a
print("b es ", b)
```

```python
a = 3
b = a
a = 'spam'
print("b es ", b)
```

·Clave: a = 'spam' no modifica el 3, sino que hace que a apunte a otro objeto. Por eso, b sigue valiendo 3.

·Key: a = ‘spam’ does not change the value of 3; rather, it causes a to point to a different object. Therefore, b still equals 3.

#### Los objetos integer no son mutables:

```python
a = 3
b = a
a = a + 2
print("b es ", b)
```

# OBJETOS MUTABLES

puedes cambiar su contenido después de crearlo, sin tener que crear un objeto nuevo.


## Listas

```python
L = [2, 3, 4]	    # A mutable object # Un objeto mutable
M = L			    # Make a reference to the same object # Crear una referencia al mismo objeto
L[0] = 24 		    # An in-place change # Un cambio in situ

print("L = ", L)    # [24, 3, 4]
print("M = ", M)    # [24, 3, 4]
```

Copia de objetos en vez de referencia:

```python
L = [2, 3, 4]
M = L[:]		    # Make a copy of L (or list(L),copy.copy (L), etc.) # Crea una copia de L (o list(L), copy.copy(L), etc.)

L[0] = 24

print("L = ", L)    # [24, 3, 4]

print("M = ", M)    # [2, 3, 4]  M is not changed. M no cambia.
```

## Referencias compartidas e igualdad
Operador == igualdad de valores.
Operador is identidad de objetos.

#### Compara el valor de dos objetos
#### 'is'
True --> Si el valor e identidad de los dos es igual

False --> Si el valor e identidad de los dos es diferente
#### '=='
True --> Si el valor del contenido de los dos es igual

False --> Si el valor del contenido de los dos es diferente

```python
L = [1, 2, 3]
M = L 			    # M and L reference the same object => M y L referencian a los mismos objetos

L == M 			    # True: Same values => operador igualdad de valores
L is M			    # True: operador identidad de objetos: compara las referencias (los punteros)

```

```python
L = [1, 2, 3]
M = [1, 2, 3]	    # M and L reference different objects
                    # «M» y «L» se refieren a objetos distintos

print(
    "L == M",
    L == M 			# True: Same values. Mismos valores.
)

print(
    "L is M",
                    # False: operador identidad de objetos: compara las identidades de cada uno pero al tener diferente identidad es falso
                    # False: object identity operator: compares the identities of each object, but since they have different identities, the result is false
    L is M
)
```

## CACHE
Sin embargo, dado que los números enteros pequeños y las cadenas se almacenan en caché y se reutilizan, esto nos indica que hacen referencia al mismo objeto.

Because small integers and strings are cached and reused, though, is tells us they reference the same single object.

```python
X = 42
Y = 42		    # Should be two different objects

X  == Y         # True
X is Y 		    # True: Same object anyhow: caching at work!!
```
