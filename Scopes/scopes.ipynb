# SCOPES
## GLOBAL SCOPE

    -Los nombres globales son variables a las que se asigna un valor en el nivel superior del archivo del módulo que las contiene.
    -Los nombres globales solo deben declararse si se les asigna un valor dentro de una función.
    -Se puede hacer referencia a los nombres globales dentro de una función sin necesidad de declararlos.

```python
x = 88 		# Global X

def func():
	x = 99 	# Local X: hides global, but we want this here

func()
print(x) #	 Prints 88: unchanged
```

En el siguiente fragmento, explica que, en este caso, el resultado es 99, ya que la función ha modificado el valor de la variable global «x».
In the following block, he explains that, in this case, the result is 99, since the function has changed the value of the global variable “x.”

```python
x = 88 			# Global X

def func():
	global x
	x = 99 		# Global X: outside def

func()
print("x =", x) # Prints 99
```

# LEGB rule
Se puede hacer referencia a nombres globales dentro de una función sin necesidad de declararlos.

```python
# Global names may be referenced within a function without being declared.
y, z = 1, 2         # Global variables in module

def all_global():
	global x        # Declare globals assigned
	x = y + z       # No need to declare y, z: LEGB rule
                    # x existe ahora en el ambito global con valor 3

all_global()
print('x = %d, y = %d, z = %d' % (x, y, z))
```

# ÁMBITOS ANIDADOS
## NESTED SCOPES

```python
x = 99                      # Global scope name: not used

def f1():
    x = 88
    def f2():               # Enclosing def local
        print("x local = ", x)    # Reference made in nested def
    f2()                    # f2 is a temporary function that lives only during the execution                               of (and is visible only to code in) the enclosing f1

f1()                        # Prints 88: enclosing def local
print("x global = ", x)     # salida: X = 88; X sigue valiendo 99
```

# Funciones de fábrica: cierres

Un cierre o una función de fábrica: el primero describe una técnica de programación funcional, mientras que el segundo hace referencia a un patrón de diseño. El objeto de función en cuestión recuerda los valores de los ámbitos circundantes, independientemente de si dichos ámbitos siguen presentes en la memoria.

## Factory Functions: Closures

A closure or a factory function, the former describing a functional programming technique, and the latter denoting a design pattern. The function object in question remembers values in enclosing scopes regardless of whether those scopes are still present in memory.

```python
def f1():
	x = 88                 # enclosing scope
	def f2():
		print("x =", x)    # Remembers x in enclosing def scope
	return f2 	           # Return f2 but don't call it => f1() devuelve el objeto funcion con                              nombre (referencia) f2

action = f1()       # Make, return function => action es ahora una función: action = f2
action() 			# Call it now: prints 88  == f2()
```

```python
def maker(n):
	def action(x):		# Make and return action
		return x ** n 	# action retains n from enclosing scope
	return action

f = maker(2)            # Pass 2 to argument n -> return x ** 2
f(3)                    # Pass 3 to x, n remembers "2" -> return 3 ** 2

f(4)
```
