# Diccionarios
Detalles del parámetro

## Sección 19.1: Introducción a los diccionarios
Un diccionario es un ejemplo de estructura de almacenamiento de pares clave-valor, también conocida como Mapping en Python. Permite guardar y recuperar elementos haciendo referencia a una clave. Como se accede a los diccionarios mediante claves, las búsquedas son muy rápidas. En las versiones de Python descritas en este capítulo, no se consideran estructuras ordenadas.
Crear un diccionario. Los diccionarios se pueden inicializar de varias maneras.
### Sintaxis literal

```python
d = {} # empty dict
d = {'key': 'value'} # dict with initial values
```

**Python 3.x**

```python
# Also unpacking one or multiple dictionaries with the literal syntax is possible
# makes a shallow copy of otherdict
d = {**otherdict}
# also updates the shallow copy with the contents of the yetanotherdict.
d = {**otherdict, **yetanotherdict}
```

### Comprensión de diccionarios

```python
d = {k:v for k,v in [('key', 'value',)]}
```

Véase también: comprensiones.
### Clase integrada dict()

```python
d = dict() # empty dict
d = dict(key='value') # explicit keyword arguments
d = dict([('key', 'value')]) # passing in a list of key/value pairs
# make a shallow copy of another dict (only possible if keys are only strings!)
d = dict(**otherdict)
```

### Modificar un diccionario
Para añadir elementos, crea una clave nueva y asígnale un valor:

```python
d['newkey'] = 42
```

También se pueden utilizar listas y diccionarios como valores:

```python
d['new_list'] = [1, 2, 3]
d['new_dict'] = {'nested_dict': 1}
```

Para eliminar un elemento, elimina su clave:

```python
del d['newkey']
```

## Sección 19.2: Evitar excepciones KeyError
Un error habitual es intentar acceder a una clave que no existe. Esto suele provocar una excepción KeyError.

```python
mydict = {}
mydict['not there']
Traceback (most recent call last):
File "<stdin>", line 1, in <module>
KeyError: 'not there'
```

Una forma de evitarlo es utilizar dict.get, que permite indicar un valor predeterminado para devolver si la clave no existe.

```python
value = mydict.get(key, default_value)
```

Devuelve mydict[key] si existe; si no, devuelve default_value. Esto no añade la clave al diccionario. Si quieres guardar ese par clave-valor, usa mydict.setdefault(key, default_value).

```python
mydict = {}
print(mydict)
# {}
print(mydict.get("foo", "bar"))
# bar
print(mydict)
# {}
print(mydict.setdefault("foo", "bar"))
# bar
print(mydict)
# {'foo': 'bar'}
```

Otra alternativa es capturar la excepción:

```python
try:
value = mydict[key]
except KeyError:
value = default_value
```

También puedes comprobar si la clave está en el diccionario:

```python
if key in mydict:
value = mydict[key]
else:
value = default_value
```

En entornos con varios hilos, la clave podría eliminarse después de comprobar que existe. Esto crea una condición de carrera y la excepción aún podría producirse. Otra opción es collections.defaultdict, una subclase de dict que utiliza default_factory para crear entradas al acceder a claves nuevas.
## Sección 19.3: Recorrer un diccionario.
Si utilizas un diccionario como iterador (por ejemplo, en un bucle for), se recorren sus claves.

```python
d = {'a': 1, 'b': 2, 'c':3}
for key in d:
print(key, d[key])
# c 3
# b 2
# a 1
```

Lo mismo sucede en una comprensión:

```python
print([key for key in d])
# ['c', 'b', 'a']
```

En Python 3, items() permite recorrer las claves y sus valores a la vez:

```python
for key, value in d.items():
print(key, value)
# c 3
# b 2
# a 1
```

El método values() permite recorrer únicamente los valores. En Python 2, keys(), values() e items() devuelven listas; también existen iterkeys(), itervalues() e iteritems(), que devuelven iteradores.

```python
for key, value in d.values():
print(key, value)
# 3
# 2
# 1
```

## Sección 19.4: Diccionarios con valores predeterminados
La biblioteca estándar proporciona defaultdict:

```python
from collections import defaultdict
d = defaultdict(int)
d['key'] # 0
d['key'] = 5
d['key'] # 5
d = defaultdict(lambda: 'empty')
d['key'] # 'empty'
d['key'] = 'full'
d['key'] # 'full'
```

Si necesitas usar la clase dict, dict.setdefault() permite crear un valor predeterminado al acceder a una clave inexistente:

```python
>>> d = {}
{}
>>> d.setdefault('Another_key', []).append("This worked!")
>>> d
```

Si añades muchos valores, setdefault() crea una nueva instancia del valor inicial (en este caso, []) cada vez que se invoca, lo que puede generar trabajo innecesario.
## Sección 19.5: Combinar diccionarios
Considera estos diccionarios:

```python
>>> fish = {'name': "Nemo", 'hands': "fins", 'special': "gills"}
>>> dog = {'name': "Clifford", 'hands': "paws", 'color': "red"}
```

Python 3.5 o posterior:

```python
fishdog = {**fish, **dog}
fishdog
{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Las claves duplicadas toman el valor del último diccionario: por ejemplo, “Clifford” sustituye a “Nemo”.
Python 3.3 o posterior

```python
>>> from collections import ChainMap
>>> dict(ChainMap(fish, dog))
{'hands': 'fins', 'color': 'red', 'special': 'gills', 'name': 'Nemo'}
```

Con esta técnica prevalece el valor del primer diccionario para una clave determinada (“Nemo” se conserva en lugar de “Clifford”). \
Python 2 y 3:

```python
>>> from itertools import chain
>>> dict(chain(fish.items(), dog.items()))
{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Esta técnica usa el último valor, igual que la combinación con **.

```python
>>> fish.update(dog)
>>> fish
{'color': 'red', 'hands': 'paws', 'name': 'Clifford', 'special': 'gills'}
```

dict.update() utiliza el segundo diccionario para sobrescribir los valores del primero.

## Sección 19.6: Acceder a claves y valores
A menudo es necesario acceder a todas las claves y valores de un diccionario, ya sea en un bucle, una comprensión de listas o como una lista.

```python
mydict = {
'a': '1','b': '2'
}
```

Para obtener las claves, utiliza keys():

```python
print(mydict.keys())
# Python2: ['a', 'b']
# Python3: dict_keys(['b', 'a'])
```

Para obtener los valores, utiliza values():

```python
print(mydict.values())
# Python2: ['1', '2']
# Python3: dict_values(['2', '1'])
```

Para obtener cada clave junto con su valor, utiliza items():

```python
print(mydict.items())
# Python2: [('a', '1'), ('b', '2')]
# Python3: dict_items([('b', '2'), ('a', '1')])
```

Nota: en las versiones de Python a las que se refiere el texto, un diccionario no está ordenado; por tanto, keys(), values() e items() no garantizan un orden. Usa sort(), sorted() u OrderedDict si necesitas controlar el orden.
Diferencia entre Python 2 y 3: en Python 3 estos métodos devuelven objetos iterables especiales, no listas, y equivalen a iterkeys(), itervalues() e iteritems() de Python 2. Pueden utilizarse como listas en muchos casos, aunque existen diferencias. Consulta PEP 3106 para más detalles.
