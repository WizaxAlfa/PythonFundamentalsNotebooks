# Diccionarios
Detalles del parámetro

## Sección 19.1: Introducción a los diccionarios
Un diccionario es un ejemplo de estructura de almacenamiento de pares clave-valor, también conocida como Mapping en Python. Permite guardar y recuperar elementos haciendo referencia a una clave. Como se accede a los diccionarios mediante claves, las búsquedas son muy rápidas. En las versiones de Python descritas en este capítulo, no se consideran estructuras ordenadas.
Crear un diccionario. Los diccionarios se pueden inicializar de varias maneras.
### Sintaxis literal

```python
d = {} # empty dict
d = {'key': 'value'} # dict with initial values
```

**Python 3.x**

```python
# Also unpacking one or multiple dictionaries with the literal syntax is possible
# makes a shallow copy of otherdict
d = {**otherdict}
# also updates the shallow copy with the contents of the yetanotherdict.
d = {**otherdict, **yetanotherdict}
```

### Comprensión de diccionarios

```python
d = {k:v for k,v in [('key', 'value',)]}
```

Véase también: comprensiones.
### Clase integrada dict()

```python
d = dict() # empty dict
d = dict(key='value') # explicit keyword arguments
d = dict([('key', 'value')]) # passing in a list of key/value pairs
# make a shallow copy of another dict (only possible if keys are only strings!)
d = dict(**otherdict)
```

### Modificar un diccionario
Para añadir elementos, crea una clave nueva y asígnale un valor:

```python
d['newkey'] = 42
```

También se pueden utilizar listas y diccionarios como valores:

```python
d['new_list'] = [1, 2, 3]
d['new_dict'] = {'nested_dict': 1}
```

Para eliminar un elemento, elimina su clave:

```python
del d['newkey']
```

## Sección 19.2: Evitar excepciones KeyError
Un error habitual es intentar acceder a una clave que no existe. Esto suele provocar una excepción KeyError.

```python
mydict = {}
mydict['not there']
Traceback (most recent call last):
File "<stdin>", line 1, in <module>
KeyError: 'not there'
```

Una forma de evitarlo es utilizar dict.get, que permite indicar un valor predeterminado para devolver si la clave no existe.

```python
value = mydict.get(key, default_value)
```

Devuelve mydict[key] si existe; si no, devuelve default_value. Esto no añade la clave al diccionario. Si quieres guardar ese par clave-valor, usa mydict.setdefault(key, default_value).

```python
mydict = {}
print(mydict)
# {}
print(mydict.get("foo", "bar"))
# bar
print(mydict)
# {}
print(mydict.setdefault("foo", "bar"))
# bar
print(mydict)
# {'foo': 'bar'}
```

Otra alternativa es capturar la excepción:

```python
try:
value = mydict[key]
except KeyError:
value = default_value
```

También puedes comprobar si la clave está en el diccionario:

```python
if key in mydict:
value = mydict[key]
else:
value = default_value
```

En entornos con varios hilos, la clave podría eliminarse después de comprobar que existe. Esto crea una condición de carrera y la excepción aún podría producirse. Otra opción es collections.defaultdict, una subclase de dict que utiliza default_factory para crear entradas al acceder a claves nuevas.
## Sección 19.3: Recorrer un diccionario.
Si utilizas un diccionario como iterador (por ejemplo, en un bucle for), se recorren sus claves.

```python
d = {'a': 1, 'b': 2, 'c':3}
for key in d:
print(key, d[key])
# c 3
# b 2
# a 1
```

Lo mismo sucede en una comprensión:

```python
print([key for key in d])
# ['c', 'b', 'a']
```

En Python 3, items() permite recorrer las claves y sus valores a la vez:

```python
for key, value in d.items():
print(key, value)
# c 3
# b 2
# a 1
```

El método values() permite recorrer únicamente los valores. En Python 2, keys(), values() e items() devuelven listas; también existen iterkeys(), itervalues() e iteritems(), que devuelven iteradores.

```python
for key, value in d.values():
print(key, value)
# 3
# 2
# 1
```

## Sección 19.4: Diccionarios con valores predeterminados
La biblioteca estándar proporciona defaultdict:

```python
from collections import defaultdict
d = defaultdict(int)
d['key'] # 0
d['key'] = 5
d['key'] # 5
d = defaultdict(lambda: 'empty')
d['key'] # 'empty'
d['key'] = 'full'
d['key'] # 'full'
```

Si necesitas usar la clase dict, dict.setdefault() permite crear un valor predeterminado al acceder a una clave inexistente:

```python
>>> d = {}
{}
>>> d.setdefault('Another_key', []).append("This worked!")
>>> d
```

Si añades muchos valores, setdefault() crea una nueva instancia del valor inicial (en este caso, []) cada vez que se invoca, lo que puede generar trabajo innecesario.
## Sección 19.5: Combinar diccionarios
Considera estos diccionarios:

```python
>>> fish = {'name': "Nemo", 'hands': "fins", 'special': "gills"}
>>> dog = {'name': "Clifford", 'hands': "paws", 'color': "red"}
```

Python 3.5 o posterior:

```python
fishdog = {**fish, **dog}
fishdog
{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Las claves duplicadas toman el valor del último diccionario: por ejemplo, “Clifford” sustituye a “Nemo”.
Python 3.3 o posterior

```python
from collections import ChainMap
dict(ChainMap(fish, dog))
{'hands': 'fins', 'color': 'red', 'special': 'gills', 'name': 'Nemo'}
```

Con esta técnica prevalece el valor del primer diccionario para una clave determinada (“Nemo” se conserva en lugar de “Clifford”). \
Python 2 y 3:

```python
from itertools import chain
dict(chain(fish.items(), dog.items()))
{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Esta técnica usa el último valor, igual que la combinación con **.

```python
fish.update(dog)
fish
{'color': 'red', 'hands': 'paws', 'name': 'Clifford', 'special': 'gills'}
```

dict.update() utiliza el segundo diccionario para sobrescribir los valores del primero.

## Sección 19.6: Acceder a claves y valores
A menudo es necesario acceder a todas las claves y valores de un diccionario, ya sea en un bucle, una comprensión de listas o como una lista.

```python
mydict = {
'a': '1','b': '2'
}
```

Para obtener las claves, utiliza keys():

```python
print(mydict.keys())
# Python2: ['a', 'b']
# Python3: dict_keys(['b', 'a'])
```

Para obtener los valores, utiliza values():

```python
print(mydict.values())
# Python2: ['1', '2']
# Python3: dict_values(['2', '1'])
```

Para obtener cada clave junto con su valor, utiliza items():

```python
print(mydict.items())
# Python2: [('a', '1'), ('b', '2')]
# Python3: dict_items([('b', '2'), ('a', '1')])
```

Nota: en las versiones de Python a las que se refiere el texto, un diccionario no está ordenado; por tanto, keys(), values() e items() no garantizan un orden. Usa sort(), sorted() u OrderedDict si necesitas controlar el orden.
Diferencia entre Python 2 y 3: en Python 3 estos métodos devuelven objetos iterables especiales, no listas, y equivalen a iterkeys(), itervalues() e iteritems() de Python 2. Pueden utilizarse como listas en muchos casos, aunque existen diferencias. Consulta PEP 3106 para más detalles.

## Sección 19.7: Acceder a los valores de un diccionario

```python
dictionary = {"Hello": 1234, "World": 5678}
print(dictionary["Hello"])
```

El código anterior imprime 1234. La cadena "Hello" es una clave: se utiliza para buscar un valor en el diccionario escribiéndola entre corchetes. El número 1234, situado después de los dos puntos, es el valor asociado a esa clave.
Buscar con una clave inexistente provoca KeyError y, si no se captura, detiene la ejecución. Para evitarlo, utiliza dictionary.get(). Si la clave no existe, devuelve None de forma predeterminada; puedes indicar un segundo valor alternativo.

```python
w = dictionary.get("whatever")
x = dictionary.get("whatever", "nuh-uh")
```

En este ejemplo, w recibe None y x recibe "nuh-uh".
## Sección 19.8: Crear un diccionario
Reglas para crear un diccionario:
• Cada clave debe ser *única*; si se repite, su valor se sobrescribe.
• Cada clave debe ser *hashable*, es decir, debe poder procesarse con la función hash; de lo contrario, se produce TypeError.
• En el contexto de las versiones descritas, las claves no tienen un orden particular.

```python
# Creating and populating it with values
stock = {'eggs': 5, 'milk': 2}
# Or creating an empty dictionary
dictionary = {}
# And populating it after
dictionary['eggs'] = 5
dictionary['milk'] = 2
# Values can also be lists
mydict = {'a': [1, 2, 3], 'b': ['one', 'two', 'three']}
# Use list.append() method to add new elements to the values list
mydict['a'].append(4)   # => {'a': [1, 2, 3, 4], 'b': ['one', 'two', 'three']}
mydict['b'].append('four')  # => {'a': [1, 2, 3, 4], 'b': ['one', 'two', 'three', 'four']}
# We can also create a dictionary using a list of two-items tuples
iterable = [('eggs', 5), ('milk', 2)]
dictionary = dict(iterables)
# Or using keyword argument:
dictionary = dict(eggs=5, milk=2)
# Another way will be to use the dict.fromkeys:
dictionary = dict.fromkeys((milk, eggs))  # => {'milk': None, 'eggs': None}
dictionary = dict.fromkeys((milk, eggs), (2, 5))  # => {'milk': 2, 'eggs': 5}
```

## Sección 19.9: Crear un diccionario ordenado
Puedes crear un diccionario que mantenga un orden determinado al recorrer sus claves. Utiliza OrderedDict del módulo collections. Al iterarlo, devuelve los elementos en el orden original de inserción.

```python
from collections import OrderedDict
d = OrderedDict()
d['first'] = 1
d['second'] = 2
d['third'] = 3
d['last'] = 4
# Outputs "first 1", "second 2", "third 3", "last 4"
for key in d:
    print(key, d[key])
```

## Sección 19.10: Desempaquetar diccionarios con el operador **
El operador ** permite pasar los pares clave-valor de un diccionario a los argumentos de una función. Este ejemplo simplificado procede de la documentación oficial:

```python
def parrot(voltage, state, action):
    print("This parrot wouldn't", action, end=' ')
    print("if you put", voltage, "volts through it.", end=' ')
    print("E's", state, "!")

d = {"voltage": "four million", "state": "bleedin' demised", "action": "VOOM"}
parrot(**d)
#This parrot wouldn't VOOM if you put four million volts through it. E's bleedin' demised !
```

Desde Python 3.5 también se puede usar esta sintaxis para combinar varios diccionarios:

```python
fish = {'name': "Nemo", 'hands': "fins", 'special': "gills"}
dog = {'name': "Clifford", 'hands': "paws", 'color': "red"}
fishdog = {**fish, **dog}
fishdog
{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Como muestra el ejemplo, las claves duplicadas toman el valor del último diccionario; por ejemplo, “Clifford” sustituye a “Nemo”.

## Sección 19.12: El constructor dict()
El constructor dict() puede crear diccionarios a partir de argumentos con nombre, de un iterable de pares clave-valor o de un diccionario junto con argumentos con nombre.

```python
dict(a=1, b=2, c=3)                   # {'a': 1, 'b': 2, 'c': 3}
dict([('d', 4), ('e', 5), ('f', 6)])  # {'d': 4, 'e': 5, 'f': 6}
dict([('a', 1)], b=2, c=3)            # {'a': 1, 'b': 2, 'c': 3}
dict({'a' : 1, 'b' : 2}, c=3)         # {'a' : 1, 'b' : 2, 'c' : 3}
```

## Sección 19.13: Ejemplo de diccionarios
Los diccionarios asocian claves con valores.

```python
car = {}
car["wheels"] = 4
car["color"] = "Red"
car["model"] = "Corvette"
```

Los valores se pueden consultar mediante sus claves.

```python
print ("Little " + car["color"] + " " + car["model"] + "!")
# This would print out "Little Red Corvette!"
```

También se pueden crear diccionarios con una sintaxis similar a JSON:

```python
car = {"wheels": 4, "color": "Red", "model": "Corvette"}
```

Se pueden recorrer los valores del diccionario:

```python
for key in car:
    print (key + ": " + car[key])
# wheels: 4
# color: Red
# model: Corvette
```

## Sección 19.14: Todas las combinaciones de valores de un diccionario

```python
options = {
    "x": ["a", "b"],
    "y": [10, 20, 30]
}
```

Supongamos que cada clave tiene una lista de valores posibles que queremos probar. Por ejemplo, se quiere explorar x="a" con y=10, y después las demás combinaciones, hasta recorrerlas todas.
El siguiente código crea una lista con todas las combinaciones posibles:

```python
import itertools
options = {
    "x": ["a", "b"],
    "y": [10, 20, 30]}
keys = options.keys()
values = (options[key] for key in keys)
combinations = [dict(zip(keys, combination)) for combination in itertools.product(*values)]
print (combinations)
```

La variable combinations contiene la siguiente lista:

```python
[{'x': 'a', 'y': 10},
{'x': 'b', 'y': 10},
{'x': 'a', 'y': 20},
{'x': 'b', 'y': 20},
{'x': 'a', 'y': 30},
{'x': 'b', 'y': 30}]
```
