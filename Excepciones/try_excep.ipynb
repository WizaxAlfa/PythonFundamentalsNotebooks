## Sintaxis de la sentencia try
La sentencia `try` debe estar acompañada de `except` o de `finally`, y el orden de las partes debe ser el siguiente.

Recuerda que en la documentación sobre líneas de comandos y código en general los símbolos `[]` indican que ese argumento es opcional y no se requiere para que la sentencia se ejecute.

Además, el metacarácter `*` indica que la sentencia `except` puede aparecer múltiples veces:

```try:
    statements
except [type [as value]]:
    statements
[except [type [as value]]:
    statements]* 
[else:
    statements] 
[finally:
    statements]

o ninguna:

try:
    statements
finally:
    statements


````python
try:
    statements                 # Ejecuta primero esta acción principal # Run this main action first
except name1:
    statements                 # Se ejecuta si se produce la excepción «nombre1» durante el bloque «try» # Run if name1 is raised during try block   
except (name2, name3):
    statements                 # Se ejecutan si se produce cualquiera de estas excepciones # Run if any of these exceptions occur
except name4 as var:
    statements                  # Se ejecutan si se lanza name4; se asigna la instancia lanzada a var # Run if name4 is raised, assign instance raised to var
except:
    statements                  # Se ejecutan para todas las demás excepciones que se produzcan # Run for all other exceptions raised
    statements
    statements
else:
    statements                  # Se ejecutan si no se ha producido ninguna excepción durante el bloque «try» # Run if no exception was raised during try block  
````

## Tabla 34-1. Formas de las cláusulas de la instrucción «try»
### Table 34-1. try statement clause forms
Forma de la cláusula  |   Interpretación
----------------------|------------------
|except:  |   Captura todos los tipos de excepción (o todos los demás).|
|except nombre:  |   Captura únicamente una excepción específica.|
|except nombre as valor:  |   Captura la excepción indicada y asigna su instancia.|
|except (nombre1, nombre2):  |   Captura cualquiera de las excepciones indicadas.|
|except (nombre1, nombre2) as valor:  |   Captura cualquier excepción de las enumeradas y asigna su instancia.|
|else:  |   Se ejecuta si no se produce ninguna excepción en el bloque try.|
|finally:  |   Este bloque se ejecuta siempre al salir.|

## 

Clause form | Interpretation
------------|---------------
|except: | Catch all (or all other) exception types.|
|except name: | Catch a specific exception only.|
|except name as value: | Catch the listed exception and assign its instance.|
|except (name1, name2): | Catch any of the listed exceptions.|
|except (name1, name2) as value: | Catch any listed exception and assign its instance.|
|else: | Run if no exceptions are raised in the try block.|
|finally: | Always perform this block on exit.|
## Ejemplo try

````python
sep = '-' * 45 + '\n'
print(sep + 'EXCEPTION RAISED AND CAUGHT')
try:
    x = 'spam'[99]
except IndexError:
    print('except run') 
finally:
    print('finally run') 
print('after run')
````

````python
print(sep + 'NO EXCEPTION RAISED')
try:
    x = 'spam'[3]
except IndexError:
    print('except run') 
finally:
    print('finally run')
print('after run')
````

````python
print(sep + 'NO EXCEPTION RAISED, WITH ELSE')
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
else:
    print('else run')
finally:
    print('finally run')
print('after run')
````

````python
print(sep + 'EXCEPTION RAISED BUT NOT CAUGHT')
try:
    x = 1 / 0
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
````
