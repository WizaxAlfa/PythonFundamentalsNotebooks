## Operaciones básicas con listas

Al tratarse de secuencias, las listas admiten muchas de las mismas operaciones que las cadenas.

len() es una función integrada, no un método de un objeto de lista.
### Basic List Operations

Because they are sequences, lists support many of the same operations as strings.

len() is a built-in function, not a method of a list object.

```python
len([1, 2, 3])
```

```python
len(['one', [2, 3], 'four'])
```

Las listas responden a los operadores + y * de forma muy similar a las cadenas de caracteres.
###### lists respond to the + and * operators much like strings.

```python
# Concatenation 
[1, 2, 3] + [4, 5, 6] 
```

```python
# Repetition
['Ni!'] * 4
```

El operador «+» funciona igual tanto con listas como con cadenas; es importante saber que espera el mismo tipo de secuencia en ambos lados; de lo contrario, se produce un error de tipo.

###### + operator works the same for lists and strings, it’s important to know that it expects the same sort of sequence on both sides, otherwise, you get a type error.


```python
str([1, 2]) + "34"    # Same as "[1, 2]" + "34"
```

```python
[1, 2] + list("34")
```

## Modificación de listas in situ

Dado que las listas son mutables, admiten operaciones que modifican un objeto de lista in situ. Es decir, todas las operaciones de esta sección modifican el objeto de lista directamente, sobrescribiendo su valor anterior sin necesidad de crear una nueva copia.

### Asignaciones de índice y de fragmentos

##### Changing Lists in Place

###### Because lists are mutable, they support operations that change a list object in place. That is, the operations in this section all modify the list object directly overwriting its former value without requiring that you make a new copy.
#### Index and slice assignments

```python
L = ['spam', 'Spam', 'SPAM!']
L[1] = 'eggs'                       # Index assignment
L 
```

```python
L[0:2] = ['eat', 'more']            # Slice assignment: delete+insert, replaces items 0,1
L                                   # ['eat', 'more', 'SPAM!']
```

```python
L = [1, 2, 3]                       # Replacement/insertion
L[1:2] = [4, 5]                     # [1, 4, 5, 3]
L
```

```python
L[1:1] = [6, 7]                     # Insertion (replace nothing)
L 
```

```python
L[1:2] = []                         # Deletion (insert nothing)
L                                   # [1, 7, 4, 5, 3]
```

```python
L = [1]
L[:0] = [2, 3, 4]                   # Insert all at :0, an empty slice at front
L
```

```python
L[len(L):] = [5, 6, 7]              #  Insert all at len(L):, an empty slice at end
L
```


## Llamadas a métodos de listas

Al igual que las cadenas, los objetos de lista de Python también admiten llamadas a métodos específicos de su tipo, muchas de las cuales modifican la lista en cuestión sobre la marcha.

Los métodos son funciones (en realidad, atributos de objetos que hacen referencia a funciones) que están asociados a objetos concretos y actúan sobre ellos.
append


### List method calls

Like strings, Python list objects also support type-specific method calls, many of which change the subject list in place.

Methods are functions (really, object attributes that reference functions) that are associated with and act upon particular objects.
append

```python
L = ['eat', 'more', 'SPAM!']        
L.append('please')                  # Append method call: add item at end
L                                   # ['eat', 'more', 'SPAM!', 'please']
```

Ten en cuenta que el método `append()` solo añade un nuevo elemento al final de la lista.

Si añades una lista a otra, la lista que añades pasa a ser un único elemento al final de la primera lista.

###### Note that the append() method only appends one new element to the end of the list.

###### If you append a list to another list, the list that you append becomes a single element at the end of the ﬁrst list.

```python
a = [1, 2, 3, 4, 5, 6, 7, 7]
b = [8, 9]
a.append(b)                     # a: [1, 2, 3, 4, 5, 6, 7, 7, [8, 9]]
a[8]
```

**sort**

```python
L.sort()                            # Sort list items ('S' < 'e')
L                                   # ['SPAM!', 'eat', 'more', 'please']
```

En «sorts», el argumento «reverse» permite ordenar en orden descendente en lugar de ascendente, y el argumento «key» proporciona una función de un argumento que devuelve el valor que se utilizará en la ordenación.

###### In sorts, the reverse argument allows sorts to be made in descending instead of ascending order, and the key argument gives a one-argument function that returns the value to be used in sorting.

```python
L = ['abc', 'ABD', 'aBe']           # Sort with mixed case
L.sort()
L                                   # ['ABD', 'aBe', 'abc']
```

```python
L = ['abc', 'ABD', 'aBe']           # L ha sido cambiada in place en la operación sort() anterior, objeto mutable!
L.sort(key=str.lower)
L                                   # ['abc', 'ABD', 'aBe']
```

```python
L = ['abc', 'ABD', 'aBe']
L.sort(key=str.lower, reverse=True) # Change sort order
L                                # ['aBe', 'ABD', 'abc']
```

**pop**

El método pop de una lista se suele utilizar junto con el método append para implementar una estructura de pila rápida del tipo «último en entrar, primero en salir» (LIFO). El final de la lista actúa como la parte superior de la pila.

###### **pop**

###### The list pop method is often used in conjunction with append to implement a quick last-in first-out (LIFO) stack structure. The end of the list serves as the top of the stack.

```python
L = []
L.append(1)                         # Push onto stack
L.append(2)                         # [1, 2]
L.pop()                             # Pop off stack
L                                   # 2
```

**del**

Dado que las listas son mutables, puedes utilizar la instrucción «del» para eliminar un elemento o una sección directamente.

###### **del**

###### Because lists are mutable, you can use the del statement to delete an item or section in place.

```python
L = ['spam', 'eggs', 'ham', 'toast']
del L[0]                                # Delete one item
L
```

## Iteración sobre una lista

Python permite utilizar un bucle «for» directamente sobre una lista:

### Iterating over a list

Python supports using a for loop directly on a list:

```python
my_list = ['foo', 'bar', 'baz']
for item in my_list:
    print(item)
```

También puedes obtener la posición de cada elemento al mismo tiempo:

You can also get the position of each item at the same time:

```python

for (index, item) in enumerate(my_list):
    print('The item in position {} is: {}'.format(index, item))
```

Otra forma de recorrer una lista en función del valor del índice:

The other way of iterating a list based on the index value:

```python
for i in range(0,len(my_list)):
    print(my_list[i])
```

## Todos y cada uno

Utiliza all() para determinar si todos los valores de un iterable dan como resultado «True».

### Any and All

Use all() to determine if all the values in an iterable evaluate to True

```python
nums = [1, [1], 'a', 1]
all(nums)
# True
```

```python
chars = ['a', 'b', 'c', 'd']
all(chars)
# True
```

La función `any()` determina si uno o más valores de un iterable dan como resultado «True».

`any()` determines if one or more values in an iterable evaluate to True

```python
nums = [1, 1, 0, 1]
any(nums)
# True
```

```python
vals = [None, 0, [], False, '']
any(vals)
# False
```

Estas funciones integradas funcionan con cualquier objeto iterable, incluidos los generadores.

These built-ins work with any iterable, including generators.

```python
vals = [1, 2, 3, 4]
any(val > 12 for val in vals)
# False
```
