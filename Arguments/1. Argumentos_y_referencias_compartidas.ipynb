# Argumentos y referencias compartidas
### Paso de objetos inmutables:

```python
def f(a):	    	# a is assigned to (references) the passed object
	a = 99 			# Changes local variable a only

b = 88		
f(b)				# a and b both reference same 88 initially
print("b = ", b)	# b is not changed
```

# Paso de objetos mutables:

```python

def changer(a, b):	    # Arguments assigned references to objects
		a = 2			# Changes local name's value only
		b[0] = 'spam' 	# Changes shared object in place
	
					
x = 1
l = [1, 2]
changer(x, l)           # Caller: Pass immutable and mutable objects
print("l = ", l)        # ['spam', 2]  # x is unchanged, l is different!
assert x == 1, "x es 1"
```

# Cómo evitar que una rutina modifique los elementos mutables:
### - Pasar una copia

```python
l = [1, 2]
changer(x, l[:]) 	# Pass a copy, so our 'l' does not change 

print("l = ", l) 
```

### - Utilizar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]
def changer(a, b):
    b = b[:]		# Copy input list so we don't impact caller
    a = 2
    b[0] = 'spam'	# Changes our list copy only

changer(x, l)
assert l == [1, 2], "l es una invariante"
print("l = ", l) 
```

### - Convertir el objeto mutable a un objeto inmutable:

```python
l = [1, 2]
def changer(a, b):	   
		a = 2			
		b[0] = 'spam'   # TypeError: 'tuple' object does not support item assignment

changer(x, tuple(l))	# Pass a tuple, so changes are errors

print("l = ", l)
assert l == [1, 2], "l es una invariante"
```

# 2 Modos especiales de coincidencia de argumentos
Hay que tener cuidado de no confundir la sintaxis especial «nombre=valor» en el encabezado de una función y en una llamada a la misma; en la llamada, significa un argumento de palabra clave que se asigna por su nombre, mientras que en el encabezado especifica un valor por defecto para un argumento opcional.

## Ejemplos de palabras clave y valores predeterminados
### 1. Comportamiento por defecto:

```python
def f(a, b, c): 
    print(a, b, c)

f(1, 2, 3)          # 1 2 3
```

### 2. Palabras clave

El orden de los argumentos de izquierda a derecha ya no importa cuando se utilizan palabras clave, ya que los argumentos se identifican por su nombre, no por su posición.
#### 2. Keywords

Left-to-right order of the arguments no longer matters when keywords are used because arguments are matched by name, not by position.

```python
def f(a, b, c): 
    print(a, b, c)

f(c=3, b=2, a=1)    # 1 2 3
```

```python
f(1, c=3, b=2)		# a gets 1 by position, b and c passed by name
                    # 1 2 3

# they make your calls a bit more self documenting
# f(name='Bob', age=40, job='dev')
```

### 3. Valores por defecto

Los valores por defecto nos permiten hacer que determinados argumentos de una función sean opcionales; si no se les pasa ningún valor, a dicho argumento se le asigna su valor por defecto antes de que se ejecute la función.

#### 3. Defaults

Defaults allow us to make selected function arguments optional; if not passed a value, the argument is assigned its default before the function runs.

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)


f(1)        # 1 2 3
f(a=1)      # 1 2 3
```

Si pasamos dos valores, solo c toma su valor por defecto, y si pasamos tres valores, no se utiliza ningún valor por defecto:

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)
    
f(1, 4)     # 1 4 3
f(1, 4, 5)	# Override defaults: 1 4 5

f(1, c=6)	# Choose defaults: 1 2 6
```

### 4. Combinación de argumentos por palabra clave y por valor predeterminado

Fíjate de nuevo en que, cuando se utilizan argumentos por palabra clave en la llamada, el orden en que aparecen los argumentos no importa; Python los identifica por su nombre, no por su posición.

El llamante debe proporcionar valores para «spam» y «eggs», pero estos pueden asignarse tanto por su posición como por su nombre.

#### 4. Combining keywords and defaults

Notice again that when keyword arguments are used in the call, the order in which the arguments are listed doesn’t matter; Python matches by name, not by position.

The caller must supply values for spam and eggs, but they can be matched by position or by name.

```python
def func(spam, eggs, toast=0, ham=0):
	print((spam, eggs, toast, ham)) 	# First 2 required

func(1, 2)							# Output: (1, 2, 0, 0)
func(1, ham=1, eggs=0)				# Output: (1, 0, 0, 1)
func(spam=1, eggs=0)				# Output: (1, 0, 0, 0)
func(toast=1, eggs=2, spam=3)		# Output: (3, 2, 1, 0)
func(1, 2, 3, 4) 					# Output: (1, 2, 3, 4)
```

### 5. Argumentos solo por palabra clave en Python 3.X

También podemos utilizar el carácter * por sí solo en la lista de argumentos para indicar que una función no acepta una lista de argumentos de longitud variable, pero que, aun así, espera que todos los argumentos que sigan al * se pasen como palabras clave.


#### 5. Python 3.X Keyword-Only Arguments

We can also use a * character by itself in the arguments list to indicate that a function does not accept a variable-length argument list but still expects all arguments following the * to be passed as keywords.

```python
def kwonly(a, *b, c):
    print(a, b, c)
kwonly(1, 2, c=3)       # 1 (2,) 3
kwonly(a=1, c=3)        # 1 () 3
kwonly(1, 2, 3)         # TypeError: kwonly() missing 1 required keyword-only argument: 'c'
```

Aún así, puedes utilizar valores por defecto para los argumentos que solo admiten palabras clave, aunque aparezcan después del * en la declaración de la función:

```python
def kwonly(a, *, b='spam', c='ham'):
	print(a, b, c
kwonly(1)               # 1 spam ham
kwonly(1, c=3)          # 1 spam 3
kwonly(a=1)             # 1 spam ham
kwonly(c=3, b=2, a=1)   # 1 2 3
kwonly(1, 2)            # TypeError: kwonly() takes 1 positional argument but 2 were given
```

```python
def kwonly(a, *, b, c='spam'):
    print(a, b, c)
kwonly(1, c='eggs')     # TypeError: kwonly() missing 1 required keyword-only argument: 'b'
```

### Ejemplos de argumentos arbitrarios

Las dos últimas extensiones de coincidencia, * y **, están diseñadas para admitir funciones que aceptan cualquier número de argumentos.

Ambas pueden aparecer tanto en la definición de la función como en una llamada a la función.
Encabezados: Recopilación de argumentos

En la definición de la función, recopila los argumentos posicionales sin coincidencia en una tupla y asigna la variable args a dicha tupla.



#### Arbitrary Arguments Examples

The last two matching extensions, * and ** , are designed to support functions that take any number of arguments.

Both can appear in either the function definition or a function call.
Headers: Collecting arguments

In the function definition, collects unmatched positional arguments into a tuple and assigns the variable args to that tuple.

```python
def f(*args):
    print(args)

f()                 # ()
f(1,)               # (1,)
f(1, 2, 3, 4)       # (1, 2, 3, 4)
```

```python
def f(**args):
    print(args)
f()                 # {}
f(a=1, b=2)         # {'a': 1, 'b': 2}
```

```python
def f(a, *pargs, **kargs):
    print(a, pargs, kargs)
f(1, 2, 3, x=1, y=2)        # 1 (2, 3) {'y': 2, 'x': 1}
```

#### Calls: Unpacking arguments

syntax in a call function it unpacks a collection of arguments

```python
def func(a, b, c, d): 
    print(a, b, c, d)

args = (1, 2)
args += (3, 4)
func(*args)         # Same as func(1, 2, 3, 4)
```

```python
def func(a, b, c, d): 
    print(a, b, c, d)
args = {'a': 1, 'b': 2, 'c': 3}
args['d'] = 4
func(**args)    # Same as func(a=1, b=2, c=3, d=4)
func(*args)
```

```python
func(*(1, 2), **{'d': 4, 'c': 3})       # Same as func(1, 2, d=4, c=3)
# 1 2 3 4
func(1, *(2, 3), **{'d': 4})            # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, c=3, *(2,), **{'d': 4})         # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
func(1, *(2, 3), d=4)                   # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, *(2,), c=3, **{'d':4})      # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
```
